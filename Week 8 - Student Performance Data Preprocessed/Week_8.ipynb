```python
paremnt_score = df.groupby('parental level of education')[score_columns].mean().sum(axis=1) / len(score_columns)
paremnt_score.plot(kind='bar', figsize=(12,6))
plt.title('Average Exam Scores by Parental Level of Education')
plt.xlabel('Parental Level of Education')
plt.ylabel('Average Score')
plt.xticks(rotation=45)
plt.show()
```
